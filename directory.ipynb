```python
import os

def print_directory_tree(startpath, max_depth=10, current_depth=0):
    """Print directory tree structure with files"""
    if current_depth > max_depth:
        return
        
    # Skip some common directories
    skip_dirs = {'.git', '__pycache__', 'node_modules', '.vscode', '.idea'}
    skip_files = {'.DS_Store', '.gitignore'}
    
    for root, dirs, files in os.walk(startpath):
        # Get current depth
        level = root.replace(startpath, '').count(os.sep)
        if level > max_depth:
            continue
            
        # Sort directories and files
        dirs.sort()
        files.sort()
        
        # Remove directories we want to skip
        dirs[:] = [d for d in dirs if d not in skip_dirs]
        
        # Print current directory
        indent = '    ' * level
        print(f"{indent}{os.path.basename(root)}/")
        
        # Print files in current directory
        subindent = '    ' * (level + 1)
        for file in files:
            if file not in skip_files:
                file_ext = os.path.splitext(file)[1]
                print(f"{subindent}{file}")
        
        # Don't go deeper if we've hit max depth
        if level >= max_depth:
            dirs.clear()

# Print the entire directory structure
print("📁 Directory Structure:")
print("=" * 60)
start_path = os.getcwd()
print(f"Root: {start_path}")
print_directory_tree(start_path)
```
